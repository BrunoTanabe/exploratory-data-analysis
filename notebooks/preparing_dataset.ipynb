## 3. Transformação dos dados

```python
# transforma a coluna 'TotalCharges' de string para float, utilizando astype
# df_contracts['TotalCharges'] = df_contracts['TotalCharges'].astype(float) # Gerou erro porque existem valores não numéricos na coluna

# transforma a coluna 'TotalCharges' de string para float, utilizando to_numeric
# df_contracts['TotalCharges'] = pd.to_numeric(df_contracts['TotalCharges']) # Gerou erro porque existem valores não numéricos na coluna

# transforma coluna 'TotalCharges' de string para float, utilizando to_numeric com erros='coerce'
df_contracts['TotalCharges'] = pd.to_numeric(df_contracts['TotalCharges'], errors='coerce') # Converte os valores não numéricos para Null
```

```python
df_contracts.info()
```

## 4. Renomeação das colunas

```python
# cria um novo dataframe com a coluna 'SeniorCitizen' como 'Above65yo', utilizando rename e dict
df_customers_new = df_customers.rename(columns={'SeniorCitizen': 'Above65yo'}) # renomeia a coluna 'SeniorCitizen' para 'Above65yo' no DataFrame df_customers
df_customers.info()
```

```python
# renomeia coluna 'SeniorCitizen' para 'Above65yo', utilizando rename e dict
df_customers.rename(columns={'SeniorCitizen': 'Above65yo'}, inplace=True)
df_customers.info()
```

```python
df_customers.columns = ['CustomerID', 'Gender', 'Above65yo', 'Partner', 'Dependents'] # renomeia todas as colunas do DataFrame df_customers
df_customers.head(5)
```

## 5. Unificando datasets

```python
print(len(df_customers), len(df_services), len(df_contracts)) # garante que o número de linhas é o mesmo em todos os dataframes
df_services.rename(columns={'customerID': 'CustomerID'}, inplace=True) # renomeia coluna 'customerID' para 'CustomerId' em df_services para ficar com o mesmo nome em todos os dataframes
```

```python
df_temp = df_customers.merge(df_services, on='CustomerID') # faz um merge entre os dataframes df_customers e df_services, utilizando a coluna 'CustomerID' como chave
df_temp.head(5)
```

```python
# Mergeando os dataframes df_temp e df_contracts, utilizando merge sem renomear a coluna 'customerID' para 'CustomerId'
df_churn_temp = df_temp.merge(df_contracts, left_on=['CustomerID'], right_on=['customerID'], how='inner')
df_churn_temp.info()
```

```python
# unificar os três dataframes ao mesmo tempo, mesmo que eles tenham nomes de colunas diferentes 
df_churn = df_customers.merge(df_services, on='CustomerID').merge(df_contracts, left_on=['CustomerID'], right_on=['customerID']) 
df_churn.info()
```

```python
# Removendo a coluna repetida 'customerID' do DataFrame df_churn
df_churn.drop(columns=['customerID'], inplace=True) # mesmo que df_churn.drop(['customerID'], axis=1, inplace=True) axis=1 é o mesmo que axis='columns'
df_churn.info()
```
